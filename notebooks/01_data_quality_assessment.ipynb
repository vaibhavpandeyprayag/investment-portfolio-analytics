## Portfolio holdings validation

### Invalid shares

```python
(portfolio_df["Shares"] <= 0).sum()
```

### Portfolio Tickers present in Stocks

```python
set(portfolio_df["Ticker"]).difference(set(stocks_df["Ticker"]))
```

```python
def all_tickers_present() -> bool:
    for ticker in portfolio_df["Ticker"].to_list():
        if (ticker not in stocks_df["Ticker"].to_list()):
            return False
    return True
all_tickers_present()
```
